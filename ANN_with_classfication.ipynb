```python
import keras
import numpy as np
import pandas as pd
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import normalize, LabelEncoder, StandardScaler, OneHotEncoder
```

```python
data=pd.read_csv("/content/airline_passenger_satisfaction.csv")
data.shape
```

```python
#removing unwanted columns
data=data.drop(columns=['ID','Ease of Online Booking', 'Gate Location'])
```

```python
data.columns
```

```python
data.isnull().sum()
```

```python
#spliting feature column and target column
x=data.drop(columns=['Satisfaction'])

#target
y=data['Satisfaction']
```

```python
num_x= x.select_dtypes(include="number").columns.tolist()
cat_x=x.select_dtypes(exclude='number').columns.tolist()
```

```python
print("columns that are numeric",num_x)
print("columns that are categorical",cat_x)
```

```python
num_transformation=Pipeline(steps=[("imputer", SimpleImputer(strategy="median")),
                                   ("scaling", StandardScaler())])
cat_transformation=Pipeline(steps=[("imputer", SimpleImputer(strategy="most_frequent")),
                                   ("onehot", OneHotEncoder(handle_unknown="ignore"))])

preprocessor= ColumnTransformer(transformers=[("num", num_transformation, num_x),
                                             ("cat", cat_transformation , cat_x)],
                               remainder="drop")
preprocessor
```

```python
le = LabelEncoder()

y = le.fit_transform(y)
print(y)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42)
print(X_train.shape)
print(X_test.shape)
```

```python
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print(X_train_processed)
#print(X_test_processed)
```

```python
#neural network modules
from keras.models import Sequential
from keras.layers import Dense, Activation
#from keras.utilts import np_utilts
import tensorflow as tf
```

```python
X_train.shape[1] #how many x columns are there
```

```python
model=Sequential()


model.add(Dense(32, input_dim=X_train_processed.shape[1], activation='relu'))
model.add(Dense(64, activation='relu'))
model.add(Dense(128, activation='relu'))
model.add(Dense(256, activation='relu'))

model.add(Dense(1, activation='sigmoid'))
```

```python
model.compile(
    loss='binary_crossentropy', #error
    optimizer='adam', #gradient decent
    metrics=['accuracy'] #accuracy
)
```

```python
#training the data
model.fit( X_train_processed,y_train,
    validation_data=( X_test_processed, y_test),
    batch_size=32,
    epochs=50)
```
